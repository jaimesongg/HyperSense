```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
import pandas as pd
from scipy.stats import spearmanr

# load data
df1 = pd.read_csv("sampen_participant_avg.csv")    
df2 = pd.read_csv("ExpertCR.csv")

# keep only needed columns
df2_small = df2[["Participant", "Hyperactivity score"]]

# merge datasets
merged = df1.merge(df2_small, on="Participant", how="inner")

print("\nSpearman Correlation: Hyperactivity vs SampEn\n")

# Acceleration SampEn
r_acc, p_acc = spearmanr(
    merged["Acc_SampEn"],
    merged["Hyperactivity score"]
)

# Jerk SampEn
r_jerk, p_jerk = spearmanr(
    merged["Jerk_SampEn"],
    merged["Hyperactivity score"]
)

print(f"Acc_SampEn: r = {r_acc:.4f}, p = {p_acc:.4f}")
print(f"Jerk_SampEn: r = {r_jerk:.4f}, p = {p_jerk:.4f}")
```

```python
# load data 
df1 = pd.read_csv("sampen_participant_avg.csv")    
df2 = pd.read_csv("ExpertCR.csv")

df2_small = df2[["Participant", "Hyperactivity score"]]

# merge
merged = df1.merge(df2_small, on="Participant", how="inner")

# drop missing values
merged = merged.dropna(subset=["Jerk_SampEn", "Hyperactivity score"])

x = merged["Jerk_SampEn"]
y = merged["Hyperactivity score"]

# single plot 
plt.figure(figsize=(7, 5))
plt.scatter(x, y)

# regression line
if len(x) > 1:
    m, b = np.polyfit(x, y, 1)
    plt.plot(x, m*x + b)

# participant labels
for xi, yi, p in zip(x, y, merged["Participant"]):
    plt.text(xi, yi, str(p), fontsize=8, ha='right', va='bottom')

plt.xlabel("Jerk_SampEn")
plt.ylabel("Hyperactivity Score")
plt.title("Jerk_SampEn vs Hyperactivity Score")

plt.tight_layout()
plt.show()
```
