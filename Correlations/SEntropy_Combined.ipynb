```python
import pandas as pd

# load data
df_sampen = pd.read_csv("combined_sampen_features.csv")

# aggregate SampEn per participant
df_sampen_avg = df_sampen.groupby("Participant")[["Acc_SampEn", "Jerk_SampEn"]].mean().reset_index()
df_sampen_avg.to_csv("sampen_participant_avg.csv", index=False)
```
